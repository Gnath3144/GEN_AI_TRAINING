# Day 06 Demo: Multimodal Image Inspection

In this lab, we:
1. Encode an image into base64.
2. Submit multimodal payloads to Vision-Language Models.

```python
import base64, io
from PIL import Image, ImageDraw

# Create a synthetic test image representing an electronics motherboard with a marked capacitor
img = Image.new('RGB', (300, 200), color=(30, 80, 40))
draw = ImageDraw.Draw(img)
draw.rectangle([40, 40, 100, 120], fill=(180, 50, 50), outline=(255, 255, 255))
draw.text((45, 70), "BURNT C12", fill=(255, 255, 255))

buf = io.BytesIO()
img.save(buf, format='JPEG')
b64_str = base64.b64encode(buf.getvalue()).decode('utf-8')
print("Generated synthetic test image base64 length:", len(b64_str))
```
