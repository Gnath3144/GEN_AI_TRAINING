# Day 08 Demo: ReAct Autonomous Agent Loop

In this lab, we:
1. Implement a ReAct step executor with internal scratchpad state.
2. Enforce loop termination boundaries.

```python
class ReActScratchpad:
    def __init__(self):
        self.steps = []
    
    def add_step(self, thought: str, action: str, observation: str):
        self.steps.append({"thought": thought, "action": action, "observation": observation})
    
    def format_history(self) -> str:
        formatted = []
        for idx, s in enumerate(self.steps, start=1):
            formatted.append(f"Step {idx}:\nThought: {s['thought']}\nAction: {s['action']}\nObservation: {s['observation']}")
        return "\n\n".join(formatted)

pad = ReActScratchpad()
pad.add_step("Check catalog price", "search('Pro Headphones')", "Found $199.99")
print(pad.format_history())
```
