## 2. Scraping and Summarizing Text
Let's fetch raw web content using BeautifulSoup and summarize it.

```python
import requests
from bs4 import BeautifulSoup

sample_html = '''
<html>
<body>
<h1>Quantum Computing Breakthrough 2026</h1>
<p>Researchers at the International Institute of Technology have unveiled a 256-qubit fault-tolerant quantum processor capable of running error-corrected topological surface codes at room-ambient cooling thresholds.</p>
<p>This innovation could slash quantum simulation runtimes for pharmaceutical drug discovery from months to hours.</p>
<nav><a href="#">Home</a> | <a href="#">Terms</a></nav>
</body>
</html>
'''

soup = BeautifulSoup(sample_html, 'html.parser')
for tag in soup(['nav', 'script', 'style']):
    tag.decompose()
clean_text = ' '.join(soup.stripped_strings)

summary = client.complete(
    f"Summarize the key announcement from this article:\n\n{clean_text}",
    system_prompt="Extract the core scientific announcement and business impact in 2 bullet points."
)
print(summary)
```
