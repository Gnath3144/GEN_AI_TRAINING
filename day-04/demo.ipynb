# Day 04 Demo: Dense Embeddings & Vector Search with ChromaDB

In this lab, we:
1. Ingest enterprise insurance documents.
2. Generate semantic embeddings.
3. Query a persistent ChromaDB vector collection.

```python
import sys, os, json
sys.path.append(os.path.abspath('..'))
import chromadb
from chromadb.utils import embedding_functions

client = chromadb.Client()
collection = client.get_or_create_collection("demo_rag")

# Add sample insurance clauses
docs = [
    "Auto Policy collision deductible is $500, reduced by $100 per incident-free year down to $0.",
    "Homeowners water damage excludes external flood and storm surge without an NFIP rider.",
    "Travel insurance provides emergency medical evacuation up to $250,000 for verified crises."
]
ids = ["doc_auto", "doc_home", "doc_travel"]

collection.add(documents=docs, ids=ids)
print(f"Indexed {collection.count()} documents in ChromaDB.")
```

## 2. Querying by Semantic Similarity

```python
query = "What happens if a hurricane causes sea water to enter my living room?"
results = collection.query(query_texts=[query], n_results=1)
print("Top Match ID:", results['ids'][0][0])
print("Retrieved Document:", results['documents'][0][0])
print("Distance:", results['distances'][0][0] if 'distances' in results else 'N/A')
```
