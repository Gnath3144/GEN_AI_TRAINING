# Day 05 Demo: Advanced RAG Evaluation & Reciprocal Rank Fusion

In this lab, we:
1. Implement Reciprocal Rank Fusion (RRF) to merge lexical and semantic retrieval.
2. Implement automated Groundedness and Context Precision checks.

```python
def reciprocal_rank_fusion(dense_ranks, bm25_ranks, k=60):
    scores = {}
    for rank, doc_id in enumerate(dense_ranks):
        scores[doc_id] = scores.get(doc_id, 0.0) + (1.0 / (k + rank + 1))
    for rank, doc_id in enumerate(bm25_ranks):
        scores[doc_id] = scores.get(doc_id, 0.0) + (1.0 / (k + rank + 1))
    
    sorted_docs = sorted(scores.items(), key=lambda x: x[1], reverse=True)
    return sorted_docs

# Sample fusion test
dense_list = ['doc_A', 'doc_B', 'doc_C']
bm25_list = ['doc_C', 'doc_A', 'doc_D']

fused = reciprocal_rank_fusion(dense_list, bm25_list)
print("Fused Ranking:", fused)
```
